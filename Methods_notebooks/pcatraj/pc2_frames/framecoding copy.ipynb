```python
import random
import os
```

```python

frames = [
"file 'p2vmdframe0001.bmp'",
"file 'p2vmdframe0002.bmp'",
"file 'p2vmdframe0003.bmp'",
"file 'p2vmdframe0004.bmp'",
"file 'p2vmdframe0005.bmp'",
"file 'p2vmdframe0006.bmp'",
"file 'p2vmdframe0007.bmp'",
"file 'p2vmdframe0008.bmp'",
"file 'p2vmdframe0009.bmp'",
"file 'p2vmdframe0010.bmp'",
"file 'p2vmdframe0011.bmp'",
"file 'p2vmdframe0012.bmp'",
"file 'p2vmdframe0013.bmp'",
"file 'p2vmdframe0014.bmp'",
"file 'p2vmdframe0015.bmp'",
"file 'p2vmdframe0016.bmp'",
"file 'p2vmdframe0017.bmp'",
"file 'p2vmdframe0018.bmp'",
"file 'p2vmdframe0019.bmp'",
"file 'p2vmdframe0020.bmp'"]
```

```python
allframes = []
for scene_start,scene_end in [(0,5),(5,10),(10,15),(15,None)]:
    # start a scene
    c1 = []
    # Which frames will be in this scene
    s1 = frames[scene_start:scene_end]

    # we first walk through the frames normally (this ensures the index is preserved for the first frames)
    for frame in c1:
        c1.append(frame)

    # Randomly shuffle and duplicate 5 times, to show movements as brownian
    for _ in range(10):
        random.shuffle(s1)
        for frame in s1:
            c1.append(frame)
    
    # Once the scene is done we append all frames to the full story
    for frame in c1:
        allframes.append(frame)
```

```python
# Write a file with the original frames in the sequence of the movie
with open("frames_2.txt","w") as f:
    for line in allframes:
        f.write(line+"\n")
```

```python
# Read the sequence and copy the files in that order but naming them as indexed in the final movie (ffmpeg uses this to format)
import shutil
with open('frames_2.txt', 'r') as f:
    for i, line in enumerate(f, start=1):
        name = line.strip().split("'")[1]
        newname = f"p2vmdframe{i:04d}.bmp"
        shutil.copy(name,f"movie/{newname}")
        
```
