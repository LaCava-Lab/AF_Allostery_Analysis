```python
import random
import os
```

```python

frames = [
"file 'vmd_frame0001.bmp'",
"file 'vmd_frame0002.bmp'",
"file 'vmd_frame0003.bmp'",
"file 'vmd_frame0004.bmp'",
"file 'vmd_frame0005.bmp'",
"file 'vmd_frame0006.bmp'",
"file 'vmd_frame0007.bmp'",
"file 'vmd_frame0008.bmp'",
"file 'vmd_frame0009.bmp'",
"file 'vmd_frame0010.bmp'",
"file 'vmd_frame0011.bmp'",
"file 'vmd_frame0012.bmp'",
"file 'vmd_frame0013.bmp'",
"file 'vmd_frame0014.bmp'",
"file 'vmd_frame0015.bmp'",
"file 'vmd_frame0016.bmp'",
"file 'vmd_frame0017.bmp'",
"file 'vmd_frame0018.bmp'",
"file 'vmd_frame0019.bmp'",
"file 'vmd_frame0020.bmp'"]
```

```python
allframes = []
for scene_start,scene_end in [(0,5),(5,10),(10,15),(15,None)]:
    # start a scene
    c1 = []
    # Which frames will be in this scene
    s1 = frames[scene_start:scene_end]

    # we first walk through the frames normally (this ensures the index is preserved for the first frames)
    for frame in c1:
        c1.append(frame)

    # Randomly shuffle and duplicate 10 times, to show movements as brownian
    for _ in range(10):
        random.shuffle(s1)
        for frame in s1:
            c1.append(frame)
    
    # Once the scene is done we append all frames to the full story
    for frame in c1:
        allframes.append(frame)
```

```python
# Write a file with the original frames in the sequence of the movie
with open("frames_2.txt","w") as f:
    for line in allframes:
        f.write(line+"\n")
```

```python
# Read the sequence and copy the files in that order but naming them as indexed in the final movie (ffmpeg uses this to format)
import shutil
with open('frames_2.txt', 'r') as f:
    for i, line in enumerate(f, start=1):
        name = line.strip().split("'")[1]
        newname = f"vmd_frame{i:04d}.bmp"
        shutil.copy(name,f"movie/{newname}")
        
```
